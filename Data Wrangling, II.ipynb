```python
import pandas as pd  # For data manipulation
import numpy as np   # For numerical operations

# Step 1: Create an 'Academic Performance' dataset
data = {
    'Student_ID': range(1, 11),
    'Math_Score': [88, 92, np.nan, 75, 95, 89, 72, 85, 90, np.nan],
    'Science_Score': [80, 85, 78, 90, np.nan, 88, 76, np.nan, 84, 91],
    'English_Score': [np.nan, 79, 85, 88, 90, 76, np.nan, 81, 87, 82],
    'Attendance': [90, 85, 92, 88, 95, 100, 80, 75, 89, 92]
}

df = pd.DataFrame(data)
print(df)

# Step 2: Handling Missing Values
# Filling missing values with the median of each column
df.fillna(df.median(), inplace=True)
print(df)

# Step 3: Detect and Handle Outliers
# Using Z-score method to detect outliers
from scipy import stats
z_scores = np.abs(stats.zscore(df))
print(z_scores)
df = df[(z_scores < 3).all(axis=1)]  # Remove rows with z-score > 3
print(df)

# Step 4: Apply Data Transformation
# Using Log Transformation on 'Attendance' to reduce skewness
df['Attendance_Log'] = np.log(df['Attendance'])

# Display final processed dataset
display(df.head())
```
