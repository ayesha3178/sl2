```python
import pandas as pd  # For data manipulation
import numpy as np   # For numerical operations

# Step 2: Load dataset from an open-source platform (Kaggle - Cancer Dataset)
# Assuming the dataset is downloaded as 'cancer_data.csv'

# Load dataset
df = pd.read_csv('Cancer_Data.csv')

# Step 3: Display first few rows
display(df.head())

# Step 4: Data Preprocessing
# Checking for missing values
print("Missing Values in Dataset:")
print(df.isnull().sum())

# Get basic statistics
display(df.describe())

# Variable descriptions
print("Column Data Types:")
print(df.dtypes)

# Check the dimensions of the dataframe
print("Shape of the dataset:", df.shape)

# Step 5: Data Formatting and Normalization
# Convert categorical target column if present
print("Before conversion:")
print(df['diagnosis'].head())
df['diagnosis'] = df['diagnosis'].map({'M': 0, 'B': 1})  # Malignant = 0, Benign = 1
print("After conversion:")
print(df['diagnosis'].head())
df['diagnosis']=df['diagnosis'].map({'M':0,'B':1})
# Normalize numerical features
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    df[col] = (df[col] - df[col].mean()) / df[col].std()3

# Step 6: Convert categorical variables into quantitative variables
# Encode any remaining categorical variables
df = pd.get_dummies(df, drop_first=True)

# Display final processed dataset
display(df.head())
```

```python
import pandas as pd  # For data manipulation
import numpy as np   # For numerical operations

# Step 1: Create an 'Academic Performance' dataset
data = {
    'Student_ID': range(1, 11),
    'Math_Score': [88, 92, np.nan, 75, 95, 89, 72, 85, 90, np.nan],
    'Science_Score': [80, 85, 78, 90, np.nan, 88, 76, np.nan, 84, 91],
    'English_Score': [np.nan, 79, 85, 88, 90, 76, np.nan, 81, 87, 82],
    'Attendance': [90, 85, 92, 88, 95, 100, 80, 75, 89, 92]
}

df = pd.DataFrame(data)
print(df)

# Step 2: Handling Missing Values
# Filling missing values with the median of each column
df.fillna(df.median(), inplace=True)
print(df)

# Step 3: Detect and Handle Outliers
# Using Z-score method to detect outliers
from scipy import stats
z_scores = np.abs(stats.zscore(df))
print(z_scores)
df = df[(z_scores < 3).all(axis=1)]  # Remove rows with z-score > 3
print(df)

# Step 4: Apply Data Transformation
# Using Log Transformation on 'Attendance' to reduce skewness
df['Attendance_Log'] = np.log(df['Attendance'])

# Display final processed dataset
display(df.head())
```
