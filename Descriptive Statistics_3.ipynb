```python
import pandas as pd
import numpy as np
import seaborn as sns


# Creating a sample dataset
data = {
    'Age_Group': ['18-25', '18-25', '26-35', '26-35', '36-45', '36-45', '46-55', '46-55', '56+', '56+'],
    'Income': [25000, 27000, 32000, 35000, 40000, 42000, 50000, 52000, 60000, 62000]
}

df = pd.DataFrame(data)
# Summary statistics grouped by Age Group
summary_stats = df.groupby('Age_Group')['Income'].agg(['mean', 'median', 'min', 'max', 'std'])
print("Summary Statistics Grouped by Age Group:\n", summary_stats)

# Load Iris dataset
iris = sns.load_dataset('iris')

# Get statistics for each species
species_list = ['setosa', 'versicolor', 'virginica']
for species in species_list:
    print(f"\nSummary Statistics for Iris-{species.capitalize()}:\n")
    print(iris[iris['species'] == species].describe())

```
